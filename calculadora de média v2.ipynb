```
# Primeiro semestre
cp1 = float(input("Nota da CP1 do primeiro semestre (0 a 100): "))
cp2 = float(input("Nota da CP2 do primeiro semestre (0 a 100): "))
cp3 = float(input("Nota da CP3 do primeiro semestre (0 a 100): "))
spr1 = float(input("Nota do Sprints 1 do primeiro semestre (0 a 100):"))
spr2 = float(input("Nota do Sprints 2 do primeiro semestre (0 a 100):"))
glob = float(input("Nota da Global Solution do primeiro semestre (0 a 100):"))

notascp = [cp1, cp2, cp3]
menor = min(notascp)

spr = (spr1 + spr2)/2

soma = cp1 + cp2 + cp3 - menor
mediacp1 = soma / 2

m1s = (0.20 * spr) + (0.20 * mediacp1) + (0.60 * glob)

# Segundo semestre
cp11 = float(input("Nota da CP1 do segundo semestre (0 a 100): "))
cp22 = float(input("Nota da CP2 do segundo semestre (0 a 100): "))
cp33 = float(input("Nota da CP3 do segundo semestre (0 a 100): "))
spr11 = float(input("Nota do Sprints 1 do segundo semestre (0 a 100):"))
spr22 = float(input("Nota do Sprints 2 do segundo semestre (0 a 100):"))
glob2 = float(input("Nota da Global Solution do segundo semestre (0 a 100): "))

notascp2 = [cp11, cp22, cp33]
menor2 = min(notascp2)

sprseg = (spr11 + spr22)/2

soma2 = cp11 + cp22 + cp33 - menor2
mediacp2 = soma2 / 2

m2s = (0.20 * sprseg) + (0.20 * mediacp2) + (0.60 * glob2)

# Média final
mf = (0.40 * m1s) + (0.60 * m2s)

# Resultado final
if mf <= 39:
    print(f"Reprovado com {mf:.2f} de nota")
elif mf >= 60:
    print(f"Aprovado com {mf:.2f} de nota")
else:
    print(f"Você está de exame com {mf:.2f} de nota")
    exame = float(input("Nota do exame (0 a 100): "))

    if exame > 120 - mf:
        print("Aprovado")
    else:
        print("Reprovado")
```
